## 1. Create lists

## 2. Modify, sort, and summarise lists

```python
#List Operations
age_list = [24, 25, 26, 27, 28]
name_list =["Ram", "Raj", "Rahul", "Rakesh", "Rajesh"]
name_list.append("Yazhini")
print(name_list)

age_list.insert(2,30)
print(age_list)

name_list.remove("Yazhini")
print(name_list)

age_list.pop()
print(age_list)

age_list.extend([29, 30, 26])
print(age_list)

age_list.sort(reverse=True)
print(age_list)

print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

## 3. Access list elements and reverse a list

```python
#Accessing List Elements
age_list = [24, 25, 26, 27, 28]
name_list =["Ram", "Raj", "Rahul", "Rakesh", "Rajesh"]
print(name_list[0])
print(name_list[-1])
print(name_list[2:5])

name_list.reverse()
print(name_list)
```

## 4. Dictionary lookup and updates

```python
#Dictionary
student_marks = {
    "Ram": 85,
    "Raj": 78,
    "Rahul": 92,
    "Rakesh": 74,
    "Rajesh": 88
}

print(student_marks)

print(student_marks["Rahul"])

student_marks["Janani"] = 80
print(student_marks)

student_marks["Raj"] = 82
print(student_marks)

print(student_marks.keys())
print(student_marks.values())
print(student_marks.items())
```

## 5. Set uniqueness, union, and intersection

```python
#Sets
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print(my_set)
#A set does not allow duplicate elements

#my_set[4] = 's'
#set does not support indexing

set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print(set1)
print(set2)

print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```
